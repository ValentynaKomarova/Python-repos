```python
#підключення до робочого файлу на google drive
conect_to_file = '/content/drive/My Drive/Colab/applications(2.0).csv'
import pandas as pd
applications = pd.read_csv(conect_to_file)
print(applications.head())
```

```python
#перевірка на наявність даних
print(applications.isnull().sum())
```

```python
#прибираю дублікати
applications.drop_duplicates('applicant_id',keep='last', inplace= True)
print(applications.head())
```

```python
#заповнюю відсутні значення в колонці External Rating нулями
applications['External Rating'].fillna(0, inplace= True)
```

```python
# заміна відсутніх значень на "Середня" в колонці Education level
applications['Education level'].fillna('Середня', inplace= True)
```

```python
#підключаю ще один робочий файл з google drive
conect_to_industries = '/content/drive/My Drive/Colab/industries(2.0).csv'
industries = pd.read_csv(conect_to_industries)
print(industries.head())
```

```python
# об'єдную 2 таблички в одину
merged = pd.merge(applications, industries, on= 'Industry', how= 'left')
print(merged.head())
```

```python
#видаляю відсутні значення
merged.dropna(subset=['Amount'], inplace=True)
print(merged['Amount'])
```

```python
#перевіряю на відсутність значень
print(merged.isnull().sum())
```

```python
#додаю до рейтинга заявника 20 балів, якщо його вік більше 35 і менше 55 років
merged['Age_between'] = ((merged['Age'] >= 35) * 20) & ((merged['Age'] <= 55) * 20)
print(merged['Age_between'].head(20))
```

```python
#перетворюю колонку з датами фотрат дати
merged['Applied at'] = pd.to_datetime(merged['Applied at'], errors='coerce')
print(merged['Applied at'].dtype)
#визначаю день тижня
merged['Date name'] = merged['Applied at'].dt.day_name()
print(merged['Date name'].head(20))
```

```python
#перевіряю унікальні записи у створеній колонці
print(merged['Date name'].unique())
```

```python
weekends = ['Saturday','Sunday']
#створюю колонку, в якій додаю 20 балів до рейтингу, якщо заявка була подана не у вихідні
merged['Rating_W'] = (~merged['Date name'].isin(['Saturday','Sunday'])) * 20
print(merged.Rating_W.head(20))
print(merged.Rating_W.sum())
```

```python
#перевіряю унікальні записи в колонці Marital status
print(merged['Marital status'].unique())
```

```python
#створюю колонку, в якій додаю 20 балів, якщо заявник одружений
merged['Status_Married'] = merged['Marital status'].str.contains('Married', case=False, na = False) *20
print(merged['Status_Married'].head(20))
```

```python
#перевіряю унікальні записи в колонці location
print(merged['Location'].unique())
```

```python
#створюю колонку, в якій додаю до рейтингу 10 балів, якщо заявник з Київа чи області
merged['Reit_Loc'] = merged['Location'].str.contains('Київ чи область', case=False, na = False) *10
print(merged['Reit_Loc'].head(10))
print(merged['Reit_Loc'].sum())
```

```python
#створюю колонку, в якій додаю 20 балів, якщо External Rating більше чи дорівнює 7
merged['Ext_Rat_plus'] = ((merged['External Rating'] >= 7) * 20)
print(merged['Ext_Rat_plus'])
```

```python
#створюю колонку, в якій віднімаю 20 балів, якщо External Rating менше чи дорівнює 2
merged['Ext_Rat_minus'] = ((merged['External Rating'] <= 2) * -20)
print(merged['Ext_Rat_minus'])
```

```python
#створюю колонку з значеннями рейтингу по всіх заданих параметрах
merged['Total_Raiting'] = merged.apply(lambda row: row['Reit_Loc'] + row['Status_Married'] + row['Rating_W'] + row['Age_between'] + row['Ext_Rat_plus'] + row['Ext_Rat_minus'] + row['Score'], axis=1)
print(merged['Total_Raiting'])
print(merged['Total_Raiting'].sum())
```

```python
merged.loc[merged['External Rating'] == 0, 'Total_Raiting'] = 0
```

```python
#прибираю з Total_Raiting значення менше 0
merged = merged[merged['Total_Raiting'] > 0]
#прибираю з Total_Raiting значення більше 100
merged = merged[merged['Total_Raiting'] < 100]
print(merged['Total_Raiting'])
print(merged['Total_Raiting'].unique())
```

```python
#дістаю номери тижів з дати
merged['week_number'] = merged['Applied at'].dt.isocalendar().week
#групую рейтинг за тижнем подачі заявки та обраховую середнє значення рейтингу
result_table = round(merged.groupby('week_number')['Total_Raiting'].mean(),2)
print(f'Середній рейтинг поданих заявок за тижнем')
print(result_table)
```
