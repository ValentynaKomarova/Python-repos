```python
# підключення до робочого файлу на Google drive
results_public = '/content/drive/My Drive/Colab/survey_results_public.csv'
import pandas as pd
df = pd.read_csv(results_public)
survey = df
print(survey.head())
```

```python
#Скільки респондентів пройшли опитування?
count_responce = df.ResponseId.count()
print(f'How many respondents were surveyed? {count_responce}')

```

```python
#фільтрую рядки без пропусків
exclusion = survey.drop(columns =['ResponseId'])
all_q = exclusion.dropna()
print(f'How many respondents answered all the questions? {all_q.shape[0]}')

```

```python
#розраховую міри центральної тенденції
median = survey['WorkExp'].median()
average = survey['WorkExp'].mean()
mode = survey['WorkExp'].mode()
print(f'What are the values ​​of the measures of central tendency for the experience of the respondents? Median = {median}, Average = {average}, Mode = {mode}')
```

```python
#рахую кількість респондентів, що працюють віддалено
count_remote = survey.loc[survey['RemoteWork'] == 'Remote'].shape[0]
print(f'How many respondents work remotely? {count_remote}')

```

```python
work_in_python = survey.LanguageHaveWorkedWith.apply(lambda work_in_python: 1 if 'python' in str(work_in_python).lower() else 0)
# Створення колонку з тими хто працює на Python в датасеті
survey['work_in_python'] = survey['LanguageHaveWorkedWith'].str.contains('Python', na=False)
#рахую кількість респондентів, які працюють на python
total_work_in_python = work_in_python.sum()
#рахую відсоток тих хто працює на пайтон від загальної кількості опитуваних
percentage_work_in_python = (total_work_in_python / count_responce) * 100
round_percentage = round(percentage_work_in_python,2)
print(f'What percentage of respondents program in Python? {round_percentage} percent')
```

```python
#рахую кількість респондентів, які навчались на онлайн курсах
learn_code = survey.LearnCode.str.contains('online courses', case=False, na = False).sum()
print(f'How many respondents learned to program using online courses? {learn_code}')
```

```python
#фільрую pyton програмістів
filtr_python = survey[survey['work_in_python']]
#видаляю рядки з Nan значеннями
filtr_python = filtr_python.dropna(subset=['ConvertedCompYearly'])
average_compensations = filtr_python.groupby('Country')['ConvertedCompYearly'].mean()
median_compensations = filtr_python.groupby('Country')['ConvertedCompYearly'].median()
ddd = 'Among respondents who program in Python, grouped by country, what is the average and median compensation amount in each country?'
print(f'{ddd} Average of compensation by countries = {average_compensations} Median of compensation by countries = {median_compensations}')

```

```python
#обираю колонки за якими виводити дані
select_columns = survey[['ResponseId', 'EdLevel', 'ConvertedCompYearly']]
sorting = select_columns.sort_values(by='ConvertedCompYearly', ascending=[False])
print(f'What are the education levels of the 5 respondents with the highest compensation?')
print(sorting.head())
```

```python
group_age = survey.groupby('Age')
total_work_in_python_by_age = group_age['work_in_python'].size()
work_in_python_by_age = group_age['work_in_python'].sum()
percentage_work_in_python_by_age = (work_in_python_by_age / total_work_in_python_by_age) * 100
round_percentage_work_in_python_by_age = round(percentage_work_in_python_by_age,2)
print(f'In each age category, what percentage of respondents program in Python?')
print(round_percentage_work_in_python_by_age)
```

```python
#фільтрую тих, хто працює віддалено
remote_work = survey[survey['RemoteWork'] == 'Remote']
comp_75_percentile = remote_work['ConvertedCompYearly'].quantile(0.75)
high_percentile_remote = remote_work[remote_work['ConvertedCompYearly'] >= (comp_75_percentile)]
main_industry = high_percentile_remote['Industry'].value_counts()
print(f'Among respondents who are in the 75th percentile for average compensation and work remotely, which industries are most prevalent?')
print(main_industry.head())
```
